```python
# Deforestation Alerts & AI Satellite Imagery (CNN)
import tensorflow as tf
from tensorflow.keras import layers, models
import numpy as np

# Simulasi Raw Data Citra Satelit yang Lebih Banyak (500 sampel citra ukuran 64x64 dengan 4 Channel)
print("Membuat raw data citra satelit tiruan...")
X_satelit = np.random.rand(500, 64, 64, 4).astype(np.float32)
y_driver = np.random.randint(0, 4, size=(500,)) # 4 Kelas: Pertanian, Tambang, Logging, Alami

# Membangun Arsitektur CNN
model = models.Sequential([
    layers.Conv2D(32, (3, 3), activation='relu', input_shape=(64, 64, 4)),
    layers.MaxPooling2D((2, 2)),
    layers.Conv2D(64, (3, 3), activation='relu'),
    layers.MaxPooling2D((2, 2)),
    layers.Flatten(),
    layers.Dense(128, activation='relu'),
    layers.Dropout(0.3),
    layers.Dense(4, activation='softmax')
])

model.compile(optimizer='adam', loss='sparse_categorical_crossentropy', metrics=['accuracy'])

# Training dengan Data yang Lebih Banyak
print("Memulai proses training model CNN...")
model.fit(X_satelit, y_driver, epochs=5, batch_size=32, validation_split=0.2)
```
