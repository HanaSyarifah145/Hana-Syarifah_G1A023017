```python
# Geographic Information System (GIS) Spasial - Update Koordinat Bengkulu
import folium
from folium.plugins import MarkerCluster, HeatMap
import pandas as pd

# Raw Data Spasial Wilayah Bengkulu (Koordinat sudah disesuaikan akurat sesuai Maps)
locations = [
    {'name': 'Universitas Bengkulu (UNIB Kandang Limun)', 'lat': -3.7589, 'lon': 102.2751, 'kategori': 'Kampus', 'intensitas': 90},
    {'name': 'Kantor Gubernur Bengkulu', 'lat': -3.8187, 'lon': 102.2882, 'kategori': 'Pemerintahan', 'intensitas': 85},
    {'name': 'Benteng Marlborough', 'lat': -3.7852, 'lon': 102.2531, 'kategori': 'Wisata Sejarah', 'intensitas': 70},
    {'name': 'Pantai Panjang', 'lat': -3.8122, 'lon': 102.2575, 'kategori': 'Wisata Alam', 'intensitas': 95},
    {'name': 'Pasar Panorama', 'lat': -3.8194, 'lon': 102.2889, 'kategori': 'Ekonomi', 'intensitas': 80},
    {'name': 'Desa Muara Maras (Seluma)', 'lat': -4.1833, 'lon': 102.6667, 'kategori': 'Lokasi KKN', 'intensitas': 60},
    {'name': 'Bandara Fatmawati Soekarno', 'lat': -3.8647, 'lon': 102.3414, 'kategori': 'Transportasi', 'intensitas': 75},
    {'name': 'Kampus UNIB 2 (Gading Cempaka)', 'lat': -3.8155, 'lon': 102.3042, 'kategori': 'Kampus', 'intensitas': 65},
    {'name': 'Pantai Tapak Paderi', 'lat': -3.7792, 'lon': 102.2468, 'kategori': 'Wisata Alam', 'intensitas': 50}
]

df_gis = pd.DataFrame(locations)

# Menggunakan tiles 'Esri.WorldImagery' (peta citra satelit gratis tanpa API key)
# Center peta difokuskan pas di sekitar Kota Bengkulu
m = folium.Map(location=[-3.8000, 102.2600], zoom_start=12, tiles='Esri.WorldImagery')

marker_cluster = MarkerCluster().add_to(m)

color_map = {
    'Kampus': 'blue',
    'Pemerintahan': 'green',
    'Wisata Sejarah': 'orange',
    'Wisata Alam': 'cadetblue',
    'Ekonomi': 'purple',
    'Lokasi KKN': 'red',
    'Transportasi': 'black'
}

for idx, row in df_gis.iterrows():
    folium.Marker(
        location=[row['lat'], row['lon']],
        popup=f"<b>{row['name']}</b><br>Kategori: {row['kategori']}<br>Skor Aktivitas: {row['intensitas']}",
        tooltip=row['name'],
        icon=folium.Icon(color=color_map.get(row['kategori'], 'blue'), icon='info-sign')
    ).add_to(marker_cluster)

heat_data = [[row['lat'], row['lon'], row['intensitas']] for idx, row in df_gis.iterrows()]
HeatMap(heat_data, min_opacity=0.4, radius=25, blur=15).add_to(m)

folium.LayerControl().add_to(m)

# Tampilkan peta di Colab
m
```
