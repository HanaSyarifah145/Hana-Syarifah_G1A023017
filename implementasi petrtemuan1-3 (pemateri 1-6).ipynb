```python
# Data Analysis & AI (CRISP-DM / Pandas)
import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler, LabelEncoder
from sklearn.impute import SimpleImputer

# Raw Data
raw_data = {
    'ID_Pelanggan': [101, 102, 103, 104, 105, 106, 107, 108, 109, 110, 111, 112, 113, 114, 115],
    'Usia': [25, 30, np.nan, 45, 22, 35, 28, 50, 41, np.nan, 29, 33, 38, 24, 47],
    'Pendapatan': [5000000, 8000000, 6000000, np.nan, 3500000, 12000000, 7500000,
                   15000000, 9200000, 6100000, np.nan, 8500000, 11000000, 4000000, 13500000],
    'Wilayah': ['Bengkulu', 'Jakarta', 'Bengkulu', 'Bandung', 'Jakarta',
                'Surabaya', 'Bengkulu', 'Jakarta', 'Bandung', 'Surabaya',
                'Bengkulu', 'Jakarta', np.nan, 'Bandung', 'Surabaya'],
    'Kategori_Pelanggan': ['Silver', 'Gold', 'Gold', 'Platinum', np.nan,
                           'Silver', 'Gold', 'Platinum', 'Silver', 'Gold',
                           'Silver', 'Platinum', 'Gold', np.nan, 'Platinum']
}

df = pd.DataFrame(raw_data)
print(f"--- Total Data Mentah: {len(df)} baris ---")
print(df.head(10))

# Data Preparation: Handling Missing Values & Scaling
imputer_num = SimpleImputer(strategy='median')
df[['Usia', 'Pendapatan']] = imputer_num.fit_transform(df[['Usia', 'Pendapatan']])

imputer_cat = SimpleImputer(strategy='most_frequent')
df[['Wilayah', 'Kategori_Pelanggan']] = imputer_cat.fit_transform(df[['Wilayah', 'Kategori_Pelanggan']])

scaler = StandardScaler()
df[['Usia_Scaled', 'Pendapatan_Scaled']] = scaler.fit_transform(df[['Usia', 'Pendapatan']])

print("\n--- Data Setelah Pra-pemrosesan (Cleaned) ---")
print(df.head(10))
```

```python
# High Performance Computing (HPC) & Parallel Computing
import multiprocessing as mp
import time

def heavy_computation(x):
    return sum([i**1.5 for i in range(x)])

if __name__ == '__main__':
    # Raw data / input list yang lebih banyak (beban komputasi besar)
    data_inputs = [5000000 + (i * 500000) for i in range(12)]

    print(f"Memproses {len(data_inputs)} tugas komputasi berat secara paralel...")

    # Eksekusi Paralel (Multi-Core CPU)
    start = time.time()
    with mp.Pool(processes=mp.cpu_count()) as pool:
        par_results = pool.map(heavy_computation, data_inputs)

    print(f"Waktu Eksekusi Paralel ({mp.cpu_count()} Cores): {time.time() - start:.4f} detik")
    print("Hasil komputasi:", par_results[:5], "...")
```

```python
# Machine Learning Klasifikasi
from sklearn.datasets import make_classification
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import classification_report, confusion_matrix

# Generate Raw Dataset Tiruan yang Lebih Besar (1000 sampel, 6 fitur)
X, y = make_classification(n_samples=1000, n_features=6, n_informative=4,
                           n_classes=2, random_state=42)

print(f"Total Sampel Data: {len(X)}")
print(f"Bentuk Fitur (X): {X.shape}, Bentuk Label (y): {y.shape}")

# Splitting Data
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state=42)

# Training Model
clf = RandomForestClassifier(n_estimators=150, random_state=42)
clf.fit(X_train, y_train)

# Evaluasi
y_pred = clf.predict(X_test)
print("\n Laporan Klasifikasi ")
print(classification_report(y_test, y_pred))
```

```python
# Deforestation Alerts & AI Satellite Imagery (CNN)
import tensorflow as tf
from tensorflow.keras import layers, models
import numpy as np

# Simulasi Raw Data Citra Satelit yang Lebih Banyak (500 sampel citra ukuran 64x64 dengan 4 Channel)
print("Membuat raw data citra satelit tiruan...")
X_satelit = np.random.rand(500, 64, 64, 4).astype(np.float32)
y_driver = np.random.randint(0, 4, size=(500,)) # 4 Kelas: Pertanian, Tambang, Logging, Alami

# Membangun Arsitektur CNN
model = models.Sequential([
    layers.Conv2D(32, (3, 3), activation='relu', input_shape=(64, 64, 4)),
    layers.MaxPooling2D((2, 2)),
    layers.Conv2D(64, (3, 3), activation='relu'),
    layers.MaxPooling2D((2, 2)),
    layers.Flatten(),
    layers.Dense(128, activation='relu'),
    layers.Dropout(0.3),
    layers.Dense(4, activation='softmax')
])

model.compile(optimizer='adam', loss='sparse_categorical_crossentropy', metrics=['accuracy'])

# Training dengan Data yang Lebih Banyak
print("Memulai proses training model CNN...")
model.fit(X_satelit, y_driver, epochs=5, batch_size=32, validation_split=0.2)
```
