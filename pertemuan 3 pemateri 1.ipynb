```python
# Human-Computer Interaction & Data Visualization
import plotly.express as px
import pandas as pd

# Raw Data Penjualan Multi-Kategori yang Lebih Banyak
data_penjualan = {
    'Bulan': ['Jan', 'Feb', 'Mar', 'Apr', 'Mei', 'Jun', 'Jul', 'Ags', 'Sep', 'Okt', 'Nov', 'Des'] * 3,
    'Kategori': ['Elektronik']*12 + ['Fashion']*12 + ['Peralatan Rumah']*12,
    'Penjualan': [120, 150, 180, 220, 200, 280, 310, 290, 340, 380, 420, 490,
                  200, 210, 250, 300, 280, 350, 390, 370, 410, 450, 500, 580,
                  90,  110, 130, 160, 150, 190, 210, 200, 230, 260, 290, 340]
}

df_viz = pd.DataFrame(data_penjualan)

# Dashboard Visualisasi Bersih (Data-Ink Ratio Optimal)
fig = px.line(
    df_viz, x='Bulan', y='Penjualan', color='Kategori',
    title='<b>Dashboard Tren Penjualan Tahunan Berdasarkan Kategori</b>',
    markers=True
)

fig.update_layout(
    plot_bgcolor='white',
    xaxis=dict(showgrid=False),
    yaxis=dict(showgrid=True, gridcolor='#eaeaea'),
    font=dict(family="Arial", size=12)
)

fig.show()
```
