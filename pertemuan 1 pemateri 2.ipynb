```python
# High Performance Computing (HPC) & Parallel Computing
import multiprocessing as mp
import time

def heavy_computation(x):
    return sum([i**1.5 for i in range(x)])

if __name__ == '__main__':
    # Raw data / input list yang lebih banyak (beban komputasi besar)
    data_inputs = [5000000 + (i * 500000) for i in range(12)]

    print(f"Memproses {len(data_inputs)} tugas komputasi berat secara paralel...")

    # Eksekusi Paralel (Multi-Core CPU)
    start = time.time()
    with mp.Pool(processes=mp.cpu_count()) as pool:
        par_results = pool.map(heavy_computation, data_inputs)

    print(f"Waktu Eksekusi Paralel ({mp.cpu_count()} Cores): {time.time() - start:.4f} detik")
    print("Hasil komputasi:", par_results[:5], "...")
```
