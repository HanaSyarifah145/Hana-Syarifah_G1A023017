```python
# Machine Learning Klasifikasi
from sklearn.datasets import make_classification
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import classification_report, confusion_matrix

# Generate Raw Dataset Tiruan yang Lebih Besar (1000 sampel, 6 fitur)
X, y = make_classification(n_samples=1000, n_features=6, n_informative=4,
                           n_classes=2, random_state=42)

print(f"Total Sampel Data: {len(X)}")
print(f"Bentuk Fitur (X): {X.shape}, Bentuk Label (y): {y.shape}")

# Splitting Data
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state=42)

# Training Model
clf = RandomForestClassifier(n_estimators=150, random_state=42)
clf.fit(X_train, y_train)

# Evaluasi
y_pred = clf.predict(X_test)
print("\n Laporan Klasifikasi ")
print(classification_report(y_test, y_pred))
```
